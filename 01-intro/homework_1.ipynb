```python
# Read CSV

df = pd.read_csv("https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv")

# Get the Length of the dataset df
len(df)
```

```python
# Get number of distinct fuel_type

df.fuel_type.nunique()
```

```python
# Get Columns with Missing Values

df.isnull().sum()
```

```python
# Get maximum fuel efficiency of cars from Asia

df[df.origin.str.lower() == 'asia'].fuel_efficiency_mpg.max()
```

```python
# Get the median value of the horsepower column => median()

df.horsepower.median()
```

```python
# Get the most frequent value of the horsepower column => mode()

df.horsepower.mode()[0]
```

```python
# Number of Missing Values Before

df.horsepower.isnull().sum()
```

```python
# Update Missing values in horsepower with the Most Frequent Value of the Column

df['horsepower'] = df['horsepower'].fillna(df['horsepower'].mode()[0])
```

```python
# Number of Missing Values After

df.horsepower.isnull().sum()
```

```python
# Get the new median value of the horsepower column => median()

df.horsepower.median()
```

```python
# Sum of weights

X = df[df.origin.str.lower()=='asia'][['vehicle_weight','model_year']].head(7).values
X
```

```python
# Transpose of X

XT = X.T
XT
```

```python
# Transpose of X * X

XTX = XT.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]

w = (XTX_inv.dot(XT)).dot(y)
np.sum(w)
```
